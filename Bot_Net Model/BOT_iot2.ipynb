```python
import pandas as pd

# Load dataset
# Read CSVs
df1 = pd.read_csv("data_12.csv", header=None, low_memory=False)
df2 = pd.read_csv("data_13.csv", header=None, low_memory=False)
df3 = pd.read_csv("data_31.csv", low_memory=False)  # Keeps headers

# Assign column names from df3 to df1 and df2
df1.columns = df3.columns
df2.columns = df3.columns

df1.columns = df1.columns.str.strip()
df2.columns = df2.columns.str.strip()
df3.columns = df3.columns.str.strip()

# Concatenate
df = pd.concat([df1, df2, df3], ignore_index=True)

# Drop useless columns (only null values)
df.drop(columns=['smac', 'dmac', 'soui', 'doui', 'sco', 'dco'], inplace=True)

# Display the first few rows
df.head()
```

```python
df['sport'] = df['sport'].fillna(df['sport'].mode()[0])
df['dport'] = df['dport'].fillna(df['dport'].mode()[0])

```
